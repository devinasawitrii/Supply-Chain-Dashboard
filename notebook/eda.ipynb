```python
import pandas as pd
df = pd.read_csv("/content/supply_chain_dataset1.csv")
df.head()
```

```python
df.info()
df.duplicated().sum()
df.describe()
```

```python
print("SKU:", df['SKU_ID'].nunique())
print("Warehouse:", df['Warehouse_ID'].nunique())
print("Supplier:", df['Supplier_ID'].nunique())
print("Region:", df['Region'].nunique())

for col in ['SKU_ID', 'Warehouse_ID', 'Supplier_ID', 'Region']:
    print(f"\n{col}:")
    print(df[col].unique())
```

```python
numeric_cols = [
    'Units_Sold',
    'Inventory_Level',
    'Supplier_Lead_Time_Days',
    'Reorder_Point',
    'Order_Quantity',
    'Unit_Cost',
    'Unit_Price',
    'Promotion_Flag',
    'Stockout_Flag',
    'Demand_Forecast'
]

for col in numeric_cols:
    print(f"\n{col}")
    print(df[col].describe())
```

```python
for col in numeric_cols:
    print(col, ":", (df[col] < 0).sum())
```
